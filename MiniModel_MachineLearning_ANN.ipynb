#Training & Testing Model Klasifikasi IRIS



## **1.Pendahuluan**

Artificial Neural Network (ANN) merupakan salah satu model pembelajaran mesin yang meniru cara kerja jaringan saraf manusia. Pada laporan ini, dibuat sebuah model ANN untuk melakukan klasifikasi data bunga Iris berdasarkan empat fitur: sepal length, sepal width, petal length, dan petal width. Dataset yang digunakan adalah IRIS.csv.

Tujuan:

*   Melakukan Preprocessing data
*   Melatih model ANN
*   Menguji Performa model menggunakan data uji
*   Melakukan prediksi pada data baru


##  **2. Import Library**

Pada tahap awal dilakukan Import library yang digunakan untuk:

*   Pandas      → Membaca dan memanipulasi data tabular
*   Numpy        → Operasi matematika dan manipulasi array (matriks)
*   Scikit-learn → membagi data (train_test_split) dan mengubah skala data (StandardScaler), serta encoding label (LabelBinarizer)
*   Tensorflow   → Framework utama untuk mengidentifikasi, mengkompilasi, dan melatih model jaringan saraf Tiruan (ANN)








```python
#import library
import pandas as pd #inport pandas untuk membaca data dalam bentuk tabel
import numpy as np #inport numpy untuk operasi matematika

#scikit-Learn: untuk membagi data dan mengubah skala angka - komputer tdk bisa baca teks
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelBinarizer

#tensorFlow: framework - membuat neural network
import tensorflow as tf
from tensorflow.keras.models import Sequential  #untuk menumpuk layer
from tensorflow.keras.layers import Dense       #layer neuron biasa (terhubung penuh)
from tensorflow.keras.layers import Dropout     #layer pemutus hubungan (pencegah hafalan)
from tensorflow.keras.optimizers import Adam    #algoritma optimasi (mekanik perbaikan bobot)
```

##  **3.Pra-Pemrosesan Data**

### **3.1 Membaca Data**

  Pandas → Digunakan untuk membaca Dataset dari file IRIS.csv.

```python
# Membaca data dari file IRIS.csv
data = pd.read_csv('IRIS.csv')
print(data) #untuk uji hasil output
```



### **3.2 Memisahkan Fitur dan Label**







*   Empat Kolom Pertama → Sebagai Input (X_Input)
*    Kolom ke-5 → sebgai label (y_output)




```python
#memisahkan data (input-x) dan label (output-y)
x_input = data.values #mengambil semua nilai pada data
print(x_input)
x_input = np.delete(x_input, 4, axis=1) #menghapus kolom ke-4 (kolom label)
print(x_input) #-- tetsing hasil data x

#mengambil kolom label
y_output = data.values[:, 4] #mengambil semua baris (:,) pada kolom ke 4
print(y_output) #-- testing hasil data y
```

### **3.3 One-Hot Encoding Label (Encoding Label)**



*   Karena model ANN tidak dapat memproses label dalam bentuk teks (Iris-setosa, dll.), label output diubah menjadi format numerik One-Hot Encoding.
*   LabelBinarizer → Label berupa teks (Setosa, Versicolor, Virginica) yang diubah menjadi angka

contoh:


1.   iris - setosa → [1. 0, 0]
2.   Iris - versicolor → [0, 1, 0]

3.   Iris - Virginica → [0, 0, 1]



   



```python
#merubah data y (teks) jadi angka (one-hot encoding)
encoder = LabelBinarizer() #fungsi untuk mengubah label (teks) menjadi one-hot encoding
y_encoded = encoder.fit_transform(y_output) #melakukan perubahan
```

### **3.4 Pembagian Data Training dan Testing**

Data dibagi menjadi set pelatihan (Training) dan pengujian (Testing) untuk evaluasi kinerja model.


*   Rasio → 80% data untuk pelatihan (X_training, y_training) dan 20% data untuk pengujian (X_testing, y_testing).
*   random_state=42 → untuk memastikan pembagian data yang sama setiap kali kode dijalankan.



```python
# pemrosesan data - memisahkan data latih dan data uji
# X_training: data latih, X_testing: data uji
# y_training: label latih, y_testing: label uji
X_training, X_testing, y_training, y_testing = train_test_split(
    x_input, y_encoded, test_size=0.2, random_state=42 # 20% (0,2) data untuk data uji. Random_state = 42 (label acak)
) #(opsional) -- bisa ubah jumlah data latih dan uji untuk coba" analisis / explore
```

### **3.5 Scaling Data**

Fitur input (X) diubah skalanya menggunakan StandardScaler (Standardisasi) agar memiliki rata-rata μ = 0 dan deviasi standar σ = 1. Ini penting untuk mempercepat konvergensi pelatihan.

- StandardScaler → digunakan untuk Standarisasi data

```python
#scaling data - merubah data ke skala standar (rata-rata 0, std 1)
scaler = StandardScaler()
X_training_scaled = scaler.fit_transform(X_training) #menghitung rata-rata & ubah data train
X_testing_scaled = scaler.transform(X_testing)       #merubah data test pakai rata-rata train
```

## **4.Pembuatan Model ANN**

Model ANN dibangun menggunakan Keras dengan struktur sebagai berikut:


*   Dense Input → dengan jumlah neorun 16, dengan layar terhubung penuh pertama. Input_shope=(4,), sesuai dengan jumlah fitur input
*   Dropout → Mematikan 20% neuron secara acak untuk mencegah *overfitting.*
*   Dense output → dengan jumlah neuron 3, sesuai dengan jumlah kelas (*Iris-setosa, Iris-versicolor, Iris-virginica*). *softmax *menghasilkan probabilitas kelas.




```python
#membuat model neural network
model = Sequential([
    #input dan hidden layer awal
    #16 jumlah neuron pada layer
    #activation = Rectified Linear Unit (ReLU)
    #input_shape = (4,): jumlah data input (x)
    Dense(16, activation='relu', input_shape=(4,)),

    #dropout layer
    #memaksa mematikan neuron sebesar 20% untuk melatih model belajar -- bisa coba" rubah besar dropout untuk explore hasil model bljr
    Dropout(0.2),

    #dropout kembali
    Dropout(0.2),

    #---area untuk hidden layer--- (kalau mau nambah hidden layer)


    #output layer
    #3: jumlah kelas output (y)
    #activation = softmax: digunakan di ouput layer untuk multi class
    Dense(3, activation='softmax')
])
```

## **4.1 Kompilasi Model**



*   Optimizer → Adam dengan learning rate α = 0.01
*   Loss Function: Categorical Crossentropy → sesuai untuk masalah klasifikasi multi-kelas dengan *One-Hot Encoding*.
*   Metrics → Accuracy (Akurasi)



```python
# aturan kompilasi model
#optimizer = Adam: algoritma optimasi
#learning_rate=0.01: kecepatan model dalam belajar
optimasi = Adam(learning_rate=0.01)

#loss='categorical_crossentropy': fungsi loss untuk multi class
model.compile(optimizer=optimasi, loss='categorical_crossentropy', metrics=['accuracy']) #metrics bisa dirubah atau ditambahkan, misal f1 score, recall, dll
```

## **5. Proses Training dan Testing**

### **5.1. Pelatihan Model**
Model dilatih menggunakan data pelatihan yang telah diskalakan (X_training_scaled, y_training).


*   Epochs → 100 kali
*   Batch Size → 16 sampel per batch sebelum bobot di update.
*   Validation Split → 20% data pelatihan digunakan sebagai set validasi internal selama proses pelatihan.



```python
# proses latihan model
print("Latihan....")
history = model.fit(
    X_training_scaled, y_training, #data latih
    validation_split=0.2, #20% data latih disimpan untuk validasi (ujian saat latihan)
    epochs=100, #jumlah latihan model
    batch_size=16, #jumlah batas data yang diproses sebelum bobot diupdate
    verbose=1 #menampilkan progress
)
```

### **5.2. Evaluasi Model**

Kinerja model diuji menggunakan data pengujian yang sama sekali belum pernah dilihat oleh model (X_testing_scaled, y_testing).

```python
# evaluasi model
print("Uji evaluasi...")
# menguji model dengan data uji
loss, accuracy = model.evaluate(X_testing_scaled, y_testing)
print(f"Akurasi pada data uji: {accuracy*100:.2f}%")
```

### **5.3. Uji Prediksi Contoh**

Dua sampel data baru diuji untuk memverifikasi kemampuan klasifikasi model:

*   Sampel 1 (Diduga Iris-setosa): Fitur: [5.1, 3.5, 1.4, 0.2]
*   Sampel 2 (Diduga Iris-versicolor): Fitur: [5.7, 2.8, 4.5, 1.32]

Setiap sampel diskalakan terlebih dahulu sebelum diprediksi.


```python
#uji contoh prediksi
#iris setosa
bunga_baru = np.array([[5.1, 3.5, 1.4, 0.2]])
bunga_baru_scaled = scaler.transform(bunga_baru) #proses scaling data baru

prediksi = model.predict(bunga_baru_scaled)
kelas_terpilih = encoder.inverse_transform(prediksi) #mengubah hasil prediksi ke label asli


#iris versicolor
bunga_baru2 = np.array([[5.7,2.8,4.5,1.32]])
bunga_baru2_scaled = scaler.transform(bunga_baru2) #proses scaling data baru

prediksi2 = model.predict(bunga_baru2_scaled)
kelas_terpilih2 = encoder.inverse_transform(prediksi2) #mengubah hasil prediksi ke label asli

print(f"======== Hasil tebakan 1 ========")
print(f"Prediksi Probabilitas: {prediksi}")
print(f"Hasil Tebakan: {kelas_terpilih[0]}\n")

print(f"======== Hasil tebakan 2 ========")
print(f"Prediksi Probabilitas: {prediksi2}")
print(f"Hasil Tebakan: {kelas_terpilih2[0]}")
```
